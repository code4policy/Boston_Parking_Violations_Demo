```python
import pandas as pd
```

```python
#number 2 ART 1
df = pd.read_csv("trellint_tickets_deid_lpn_2024.csv")
```

```python
#number 4 ART 1
canonical_cols = [
    "ticket_number",
    "ticket_issue_date",
    "ticket_issue_time",
    "violation_desc_long",
    "deid_lpn",
    "latitude",
    "longitude",
    "street_no",
    "street_name",
    "street_direction",
    "street_suffix",
    "location",
]

canon = df[canonical_cols].copy()

```

```python
#number 5 ART 1
import geopandas as gpd

boston = gpd.read_file("city_of_boston_outline_boundary_water_excluded.json")

boston.shape
boston.geometry.type
```

```python
#number 6 ART 1
import geopandas as gpd

# 1) Build point GeoDataFrame from canon (no sampling)
gdf = gpd.GeoDataFrame(
    canon.copy(),
    geometry=gpd.points_from_xy(canon["longitude"], canon["latitude"]),
    crs="EPSG:4326",
)

# 2) Strict invalid coord drops (unambiguous)
m_valid = (
    gdf["latitude"].notna() & gdf["longitude"].notna() &
    (gdf["latitude"] != 0) & (gdf["longitude"] != 0)
)
gdf_valid = gdf[m_valid].copy()

# 3) Boston AOI polygon (MultiPolygon -> single geometry)
aoi = boston.geometry.iloc[0]   # MultiPolygon

# 4) Point-in-polygon filter (keep only inside Boston)
inside = gdf_valid.within(aoi)
canon_clean = gdf_valid.loc[inside].drop(columns=["geometry"]).copy()

# 5) Audit
print("canon rows:", f"{len(canon):,}")
print("after drop NaN/0 coords:", f"{len(gdf_valid):,}")
print("inside Boston AOI:", f"{len(canon_clean):,}")
print("dropped by AOI:", f"{len(gdf_valid) - len(canon_clean):,}")
```

```python
#number 8 ART 1
canon_clean["ticket_number"] = canon_clean["ticket_number"].astype("string")
canon_clean.to_parquet("tickets_2024_canonical.parquet", index=False)
```

```python
#number 9 ART 2 -ENRICHED (DUCKDB ATTACH CANONINAL)
import duckdb

con = duckdb.connect()

con.execute("""
    CREATE VIEW tickets AS
    SELECT *
    FROM 'tickets_2024_canonical.parquet'
""")
```

```python
#number 10 ART 2 ENRICHED BUILD TABLE
con.execute("""
CREATE OR REPLACE TABLE tickets_enriched AS
SELECT
  *,
  CAST(SUBSTR(ticket_issue_time, 1, 2) AS INTEGER) AS issue_hour,
  CAST(SUBSTR(ticket_issue_time, 4, 2) AS INTEGER) AS issue_minute,
  EXTRACT(DOW   FROM CAST(ticket_issue_date AS DATE)) AS issue_dow,
  EXTRACT(MONTH FROM CAST(ticket_issue_date AS DATE)) AS issue_month
FROM tickets
""")
```

```python
#number 11
print("canonical rows:", con.execute("SELECT COUNT(*) FROM tickets").fetchone()[0])
print("enriched rows :", con.execute("SELECT COUNT(*) FROM tickets_enriched").fetchone()[0])
print("bad hours     :", con.execute("SELECT COUNT(*) FROM tickets_enriched WHERE issue_hour NOT BETWEEN 0 AND 23").fetchone()[0])
print("bad minutes   :", con.execute("SELECT COUNT(*) FROM tickets_enriched WHERE issue_minute NOT BETWEEN 0 AND 59").fetchone()[0])

con.execute("""
SELECT issue_hour, COUNT(*) AS n
FROM tickets_enriched
GROUP BY 1
ORDER BY 1
LIMIT 10
""").fetchdf()
```

```python
#number 12 ENRICHED WRITE PARQUET
con.execute("COPY tickets_enriched TO 'tickets_2024_enriched.parquet' (FORMAT PARQUET)")
print("Wrote: tickets_2024_enriched.parquet")
```

```python

# 13 ART 3 
con.execute("""
CREATE OR REPLACE TABLE agg_map_grid_day AS
SELECT
  ROUND(latitude / 0.0005) * 0.0005 AS lat_bin,
  ROUND(longitude / 0.0005) * 0.0005 AS lon_bin,
  violation_desc_long,
  CAST(ticket_issue_date AS DATE) AS ticket_date,
  COUNT(*) AS ticket_count
FROM tickets_enriched
GROUP BY 1,2,3,4
""")
```

```python
#14 ART 4
con.execute("""
CREATE OR REPLACE TABLE agg_offenders_day AS
SELECT
  deid_lpn,
  CAST(ticket_issue_date AS DATE) AS ticket_date,
  violation_desc_long,
  COUNT(*) AS ticket_count
FROM tickets_enriched
WHERE deid_lpn IS NOT NULL
GROUP BY 1,2,3
""")
```

```python
#38 phase 6.1
# Global filters (UI will set these)
DATE_START = "2024-01-01"   # inclusive, string 'YYYY-MM-DD'
DATE_END   = "2024-12-31"   # inclusive, string 'YYYY-MM-DD'

# Violation filter (UI multi-select)
# Empty list means "ALL violations"
VIOLATIONS = []             # e.g. ["UNPAID METER", "EXPIRED METER"]

# Top-N controls (UI slider/input)
TOP_N = 20
```

```python
#39 phase 6.2
def q_map(con, date_start, date_end, violations):
    if violations:
        placeholders = ",".join(["?"] * len(violations))
        sql = f"""
        SELECT
          lat_bin,
          lon_bin,
          SUM(ticket_count) AS total_count,
          SUM(ticket_count) / SUM(day_weight) AS avg_per_day
        FROM agg_map_grid_day_norm
        WHERE ticket_date BETWEEN ? AND ?
          AND violation_desc_long IN ({placeholders})
        GROUP BY 1,2
        """
        params = [date_start, date_end] + violations
    else:
        sql = """
        SELECT
          lat_bin,
          lon_bin,
          SUM(ticket_count) AS total_count,
          SUM(ticket_count) / SUM(day_weight) AS avg_per_day
        FROM agg_map_grid_day_norm
        WHERE ticket_date BETWEEN ? AND ?
        GROUP BY 1,2
        """
        params = [date_start, date_end]

    return con.execute(sql, params).fetchdf()
```

```python
#40 phase 6.2
df_map = q_map(con, DATE_START, DATE_END, VIOLATIONS)
df_map.head()
```

```python
#41 phase 6.3
def q_top_offenders(con, date_start, date_end, violations, top_n, metric="total"):
    top_n = int(top_n)
    # metric: "total" or "avg_per_day"
    score_expr = "SUM(ticket_count)" if metric == "total" else "SUM(ticket_count) / SUM(day_weight)"

    if violations:
        placeholders = ",".join(["?"] * len(violations))
        sql = f"""
        SELECT
          deid_lpn,
          {score_expr} AS score,
          SUM(ticket_count) AS total_tickets,
          SUM(day_weight) AS active_days
        FROM agg_offenders_day_norm
        WHERE ticket_date BETWEEN ? AND ?
          AND violation_desc_long IN ({placeholders})
        GROUP BY 1
        ORDER BY score DESC
        LIMIT ?
        """
        params = [date_start, date_end] + violations + [top_n]
    else:
        sql = f"""
        SELECT
          deid_lpn,
          {score_expr} AS score,
          SUM(ticket_count) AS total_tickets,
          SUM(day_weight) AS active_days
        FROM agg_offenders_day_norm
        WHERE ticket_date BETWEEN ? AND ?
        GROUP BY 1
        ORDER BY score DESC
        LIMIT ?
        """
        params = [date_start, date_end, top_n]

    return con.execute(sql, params).fetchdf()
```

```python
# 42 phase 6.3
df_top_off = q_top_offenders(con, DATE_START, DATE_END, VIOLATIONS, TOP_N, metric="total")
df_top_off.head(10)
```

```python
#43 phase 6.3
df_top_off_norm = q_top_offenders(con, DATE_START, DATE_END, VIOLATIONS, TOP_N, metric="avg_per_day")
df_top_off_norm.head(10)
```

```python
# 44 phase 6.4 
SELECTED_OFFENDER = df_top_off.loc[0, "deid_lpn"]  # UI will set this from a click
SELECTED_OFFENDER
```

```python
#45 phase 6.4
def q_offender_violation_breakdown(con, deid_lpn, date_start, date_end, violations=None):
    deid_lpn = int(deid_lpn)  # <-- ADD THIS LINE
    violations = violations or []

    if violations:
        placeholders = ",".join(["?"] * len(violations))
        sql = f"""
        SELECT
          violation_desc_long,
          SUM(ticket_count) AS tickets
        FROM agg_offender_violation_day
        WHERE deid_lpn = ?
          AND ticket_date BETWEEN ? AND ?
          AND violation_desc_long IN ({placeholders})
        GROUP BY 1
        ORDER BY tickets DESC
        """
        params = [deid_lpn, date_start, date_end] + violations
    else:
        sql = """
        SELECT
          violation_desc_long,
          SUM(ticket_count) AS tickets
        FROM agg_offender_violation_day
        WHERE deid_lpn = ?
          AND ticket_date BETWEEN ? AND ?
        GROUP BY 1
        ORDER BY tickets DESC
        """
        params = [deid_lpn, date_start, date_end]

    return con.execute(sql, params).fetchdf()
```

```python
#45 phase 6.4
df_off_v = q_offender_violation_breakdown(con, SELECTED_OFFENDER, DATE_START, DATE_END, VIOLATIONS)
df_off_v.head(20)
```

```python
#46 phase 6.4 
def q_offender_hour_profile(con, deid_lpn, date_start, date_end):
    deid_lpn = int(deid_lpn)  # avoid numpy.int64 binding issues

    sql = """
    SELECT
      issue_hour,
      SUM(ticket_count) AS tickets
    FROM agg_offender_hour_day
    WHERE deid_lpn = ?
      AND ticket_date BETWEEN ? AND ?
    GROUP BY 1
    ORDER BY 1
    """
    return con.execute(sql, [deid_lpn, date_start, date_end]).fetchdf()
```

```python
#47 phase 6.4
df_off_h = q_offender_hour_profile(con, SELECTED_OFFENDER, DATE_START, DATE_END)
df_off_h
```

```python
#48 phase 6.4
def q_offender_map_grid(con, deid_lpn, date_start, date_end, cell=0.0005):
    deid_lpn = int(deid_lpn)  # avoid numpy.int64 binding issues
    cell = float(cell)

    sql = f"""
    SELECT
      ROUND(latitude / {cell}) * {cell} AS lat_bin,
      ROUND(longitude / {cell}) * {cell} AS lon_bin,
      COUNT(*) AS tickets
    FROM tickets_enriched
    WHERE deid_lpn = ?
      AND ticket_issue_date BETWEEN ? AND ?
    GROUP BY 1,2
    """
    return con.execute(sql, [deid_lpn, date_start, date_end]).fetchdf()
```

```python
#49 phase 6.4
df_off_map = q_offender_map_grid(con, SELECTED_OFFENDER, DATE_START, DATE_END)
df_off_map.head()
```

```python
#50 phase 6.5
def q_top_streets(con, date_start, date_end, violations, top_n, metric="total"):
    top_n = int(top_n)
    score_expr = "SUM(ticket_count)" if metric == "total" else "SUM(ticket_count) / SUM(day_weight)"

    if violations:
        placeholders = ",".join(["?"] * len(violations))
        sql = f"""
        SELECT
          street_key,
          {score_expr} AS score,
          SUM(ticket_count) AS total_tickets,
          SUM(day_weight) AS active_days
        FROM agg_streets_day_norm
        WHERE ticket_date BETWEEN ? AND ?
          AND violation_desc_long IN ({placeholders})
        GROUP BY 1
        ORDER BY score DESC
        LIMIT ?
        """
        params = [date_start, date_end] + violations + [top_n]
    else:
        sql = f"""
        SELECT
          street_key,
          {score_expr} AS score,
          SUM(ticket_count) AS total_tickets,
          SUM(day_weight) AS active_days
        FROM agg_streets_day_norm
        WHERE ticket_date BETWEEN ? AND ?
        GROUP BY 1
        ORDER BY score DESC
        LIMIT ?
        """
        params = [date_start, date_end, top_n]

    return con.execute(sql, params).fetchdf()
```

```python
#51 
df_top_st = q_top_streets(con, DATE_START, DATE_END, VIOLATIONS, TOP_N, metric="total")
df_top_st.head(10)
```
